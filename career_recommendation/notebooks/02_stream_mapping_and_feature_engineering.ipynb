# Stream Mapping and Feature Engineering

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/processed/cleaned_career_dataset.csv')
print(df.shape)
df.head()
```

```python
possible_target_cols = [c for c in df.columns if 'career' in c or 'job' in c or 'profession' in c]
possible_target_cols
```

```python
target_col = possible_target_cols[0]
df[target_col] = df[target_col].astype(str).str.lower().str.strip()
```

```python
pcm_keywords = ['engineer','software','developer','architect','scientist']
pcb_keywords = ['doctor','medical','nurse','biologist']
commerce_math_keywords = ['accountant','finance','banker','economist']
commerce_without_math_keywords = ['business','marketing','sales','entrepreneur']
arts_keywords = ['lawyer','journalist','teacher','writer','designer']

def map_stream(career):
    career = str(career).lower()

    if any(k in career for k in pcm_keywords):
        return 'PCM'
    elif any(k in career for k in pcb_keywords):
        return 'PCB'
    elif any(k in career for k in commerce_math_keywords):
        return 'Commerce with Maths'
    elif any(k in career for k in commerce_without_math_keywords):
        return 'Commerce without Maths'
    elif any(k in career for k in arts_keywords):
        return 'Arts / Humanities'
    else:
        return 'Other'

df['recommended_stream'] = df[target_col].apply(map_stream)

df['recommended_stream'].value_counts()
```

```python
df = df[df['recommended_stream'] != 'Other']
```

```python
df.to_csv('../data/processed/stream_mapped_career_dataset.csv', index=False)
print('Saved stream mapped dataset')
```
